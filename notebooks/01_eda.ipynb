```python
import pandas as pd
df = pd.read_csv("C:\\Users\\Mi\\OneDrive\\Desktop\\ml-insight-platform\\data\\raw\\telco_churn.csv.csv")
df.shape
df.dtypes
df.head()
df.isnull().sum()
df["Churn"].value_counts(normalize=True)  # class balance - important, churn is usually imbalanced
```

```python
df.groupby("Churn")["MonthlyCharges"].mean()
df.groupby("Churn")["tenure"].mean()
df["Contract"].value_counts()
pd.crosstab(df["Contract"], df["Churn"], normalize="index")
```

```python
missing = df.isnull().sum()

print("\nMissing values:")
print(missing[missing > 0])
```

```python
print("TotalCharges dtype:", df["TotalCharges"].dtype)

# Check blank/empty values
blank_totalcharges = df["TotalCharges"].astype(str).str.strip().eq("").sum()

print("Blank TotalCharges values:", blank_totalcharges)
df[df["TotalCharges"].astype(str).str.strip() == ""]
```

```python
print(df["Churn"].value_counts())
print("\nPercentage:")
print(df["Churn"].value_counts(normalize=True) * 100)
```

```python
contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

print(contract_churn.round(2))
```

```python
df["tenure_group"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=["0-12 months", "13-24 months", "25-48 months", "49-72 months"]
)

tenure_churn = pd.crosstab(
    df["tenure_group"],
    df["Churn"],
    normalize="index"
) * 100

print(tenure_churn.round(2))
```

```python
categorical_cols = [
    "Contract",
    "PaymentMethod",
    "InternetService",
    "TechSupport",
    "OnlineSecurity",
    "SeniorCitizen",
    "Partner",
    "Dependents"
]

for col in categorical_cols:
    print(f"\n{'='*50}")
    print(f"{col}")
    print(pd.crosstab(
        df[col],
        df["Churn"],
        normalize="index"
    ).round(2))
```

```python
# 1. Duplicate check - sanity check on customerID uniqueness
print("Duplicate customerIDs:", df["customerID"].duplicated().sum())

# 2. Numeric feature correlation - helps you spot multicollinearity before modeling
numeric_cols = ["tenure", "MonthlyCharges"]
df["TotalCharges_clean"] = pd.to_numeric(df["TotalCharges"].replace(" ", "0"), errors="coerce")
numeric_cols.append("TotalCharges_clean")

print(df[numeric_cols].corr())

# 3. Check redundancy across the "add-on service" columns
# (OnlineSecurity, OnlineBackup, DeviceProtection, TechSupport, StreamingTV, StreamingMovies)
addon_cols = ["OnlineSecurity", "OnlineBackup", "DeviceProtection",
              "TechSupport", "StreamingTV", "StreamingMovies"]
for col in addon_cols:
    print(col, ":", df[col].unique())

# 4. Control test - a feature that SHOULDN'T matter much, to validate your read of the data
print(pd.crosstab(df["gender"], df["Churn"], normalize="index").round(2))

# 5. MonthlyCharges distribution by churn boxplot, not just mean
import matplotlib.pyplot as plt
df.boxplot(column="MonthlyCharges", by="Churn")
plt.title("Monthly Charges by Churn")
plt.suptitle("")
plt.show()

# 6. Interaction: does tenure matter LESS once you control for contract type?
# (important because contract type and tenure are likely correlated with each other too)
pd.crosstab(df["Contract"], df["tenure_group"], normalize="index").round(2)
```
